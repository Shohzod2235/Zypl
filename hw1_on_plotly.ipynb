````python

# Пример данных (замените на запрос к вашей базе данных):
# SQL запрос: SELECT billing_country, total FROM invoice;
data = {
    "billing_country": ["USA", "Canada", "Germany", "France", "USA", "Germany", "Canada"],
    "total": [15.5, 20.3, 10.8, 25.4, 30.2, 12.7, 18.6]
}
df = pd.DataFrame(data)

# Создание коробчатой диаграммы
fig = px.box(
    df, 
    x="billing_country", 
    y="total", 
    title="Распределение дохода по странам", 
    labels={"billing_country": "Страна", "total": "Доход"}
)
fig.show()
````

````python

# Пример данных (замените на запрос к вашей базе данных):
# SQL запрос: SELECT DATE_TRUNC('month', invoice_date) AS month, SUM(total) AS monthly_income 
# FROM invoice WHERE EXTRACT(YEAR FROM invoice_date) = 2025 
# GROUP BY month ORDER BY month;

data = {
    "month": ["2025-01", "2025-02", "2025-03", "2025-04", "2025-05", "2025-06"],
    "monthly_income": [1500, 1800, 1700, 2000, 1900, 2100]
}
df = pd.DataFrame(data)

# Построение линейчатой диаграммы
fig = px.line(
    df, 
    x="month", 
    y="monthly_income", 
    title="Изменение дохода по месяцам (2025 год)", 
    labels={"month": "Месяц", "monthly_income": "Доход"}
)
fig.show()
````

````python

# Пример данных (замените на запрос к вашей базе данных):
# SQL запрос: SELECT 
# CASE 
#    WHEN billing_country = 'USA' THEN 'USA' 
#    ELSE 'Rest of the World' 
# END AS region, 
# SUM(total) AS sales 
# FROM invoice GROUP BY region;

data = {
    "region": ["USA", "Rest of the World"],
    "sales": [12000, 8000]
}
df = pd.DataFrame(data)

# Построение круговой диаграммы
fig = px.pie(
    df, 
    names="region", 
    values="sales", 
    title="Соотношение продаж: США и остальной мир"
)
fig.show()
````
