```python
# задача 1
import math

total_cards = 52

aces = 4

cards_to_draw = 2


def combinations(n, k):
    return math.factorial(n) // (math.factorial(k) * math.factorial(n - k))


favorable_outcomes = combinations(aces, cards_to_draw)


total_outcomes = combinations(total_cards, cards_to_draw)


probability = favorable_outcomes / total_outcomes

print(f"Вероятность вытащить два туза: {probability:.6f}")
```

```python
# задача 2 
P_L = 4 / 5 
P_not_L = 1 - P_L  
P_O_given_L = 0.7 
P_O_given_not_L = 0.1  

P_O = P_O_given_L * P_L + P_O_given_not_L * P_not_L

print(f"Вероятность того, что друг опоздает на работу: {P_O:.2f}")
```

```python
# задача 3 
total_balls = 12
green_balls = 4
red_balls = 5

# Вероятности для двух случаев
P1 = (green_balls / total_balls) * (red_balls / (total_balls - 1))
P2 = (red_balls / total_balls) * (green_balls / (total_balls - 1))

# Общая вероятность
P = P1 + P2

print(f"Вероятность вытянуть зелёный и красный в любом порядке: {P:.3f}")
```

```python
# задача 4
P_A = 0.02  
P_not_A = 0.98  
P_B_given_A = 0.975  
P_B_given_not_A = 0.01  


P_A_given_B = (P_B_given_A * P_A) / (
    P_B_given_A * P_A + P_B_given_not_A * P_not_A
)

print(f"Вероятность: {P_A_given_B:.4f}")
```

```python
# задача 5
P_A = 0.03  
P_not_A = 0.97  
P_B_given_A = 0.98  
P_B_given_not_A = 0.01  


P_B = P_B_given_A * P_A + P_B_given_not_A * P_not_A


P_A_given_B = (P_B_given_A * P_A) / P_B

print(f"Вероятность включения индикатора: {P_B:.4f}")
print(f"Вероятность перегрева при включённом индикаторе: {P_A_given_B:.4f}")
```
