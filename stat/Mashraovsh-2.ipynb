```python
#5


from math import comb

team_size = 13
# Количество выбранных люде
chosen_people = 3

# Общее количество способов выбрать 3 человека из 13
total_combinations = comb(team_size, chosen_people)

# Количество способов, в которых конкретный человек (вы) попадает в тройку
ways_with_you = comb(team_size - 1, chosen_people - 1)

# Вероятность быть выбранным
probability = ways_with_you / total_combinations

print(f"Вероятность быть выбранным на конференцию: {probability:.4f}")
```
