```python
import numpy as np
import scipy.stats as stats

alpha = 0.05  # Уровень значимости 5%

# ===============================
# Задача 1: Проверка срока службы лампочек
# ===============================
n1 = 50
mu1 = 2000  # Заявленный срок службы
sample_mean1 = 1980  # Среднее из выборки
std1 = 66  # Стандартное отклонение выборки

# Односторонний t-тест (проверка, меньше ли средний срок службы)
t_stat1 = (sample_mean1 - mu1) / (std1 / np.sqrt(n1))
p_value1 = stats.t.cdf(t_stat1, df=n1-1)  # Левая сторона

print("Задача 1:")
print(f"t-статистика: {t_stat1:.3f}, p-значение: {p_value1:.3f}")
if p_value1 < alpha:
    print("Отвергаем H0: Средний срок службы меньше заявленного.")
else:
    print("Не отвергаем H0: Недостаточно доказательств, что срок службы меньше.")


# ===============================
# Задача 2: Проверка фасовочной машины
# ===============================
n2 = 100
mu2 = 1000  # Ожидаемый вес
sample_mean2 = 1003  # Среднее из выборки
std2 = 5  # Известное стандартное отклонение

# Односторонний z-тест (проверка, больше ли средний вес)
z_stat2 = (sample_mean2 - mu2) / (std2 / np.sqrt(n2))
p_value2 = 1 - stats.norm.cdf(z_stat2)  # Правая сторона

print("\nЗадача 2:")
print(f"z-статистика: {z_stat2:.3f}, p-значение: {p_value2:.3f}")
if p_value2 < alpha:
    print("Отвергаем H0: Средний вес мешков больше 1000г.")
else:
    print("Не отвергаем H0: Недостаточно доказательств, что фасовка превышает норму.")


# ===============================
# Задача 3: Проверка срока жизни котов
# ===============================
n3 = 40
mu3 = 15.7  # Ожидаемая средняя продолжительность жизни
sample_mean3 = 16.5  # Среднее из выборки
std3 = 1.6  # Стандартное отклонение выборки

# Односторонний t-тест (проверка, больше ли средняя продолжительность жизни)
t_stat3 = (sample_mean3 - mu3) / (std3 / np.sqrt(n3))
p_value3 = 1 - stats.t.cdf(t_stat3, df=n3-1)  # Правая сторона

print("\nЗадача 3:")
print(f"t-статистика: {t_stat3:.3f}, p-значение: {p_value3:.3f}")
if p_value3 < alpha:
    print("Отвергаем H0: Средний срок жизни котов больше 15.7 лет.")
else:
    print("Не отвергаем H0: Недостаточно доказательств, что коты живут дольше.")


# ===============================
# Задача 4: Проверка веса новорожденных
# ===============================
n4 = 49
mu4 = 3.40  # Ожидаемый средний вес
sample_mean4 = 3.29  # Среднее из выборки
std4 = 0.45  # Известное стандартное отклонение

# Односторонний z-тест (проверка, меньше ли средний вес)
z_stat4 = (sample_mean4 - mu4) / (std4 / np.sqrt(n4))
p_value4 = stats.norm.cdf(z_stat4)  # Левая сторона

print("\nЗадача 4:")
print(f"z-статистика: {z_stat4:.3f}, p-значение: {p_value4:.3f}")
if p_value4 < alpha:
    print("Отвергаем H0: Средний вес новорожденных меньше 3.40 кг.")
else:
    print("Не отвергаем H0: Недостаточно доказательств, что вес новорожденных отличается.")


# ===============================
# Задача 5: Проверка влияния лекарства на IQ
# ===============================
n5 = 30
mu5 = 100  # Средний IQ в популяции
sample_mean5 = 140  # Средний IQ в выборке
std5 = 15  # Известное стандартное отклонение

# Двусторонний z-тест (проверка, отличается ли средний IQ от 100)
z_stat5 = (sample_mean5 - mu5) / (std5 / np.sqrt(n5))
p_value5 = 2 * (1 - stats.norm.cdf(abs(z_stat5)))  # Двусторонний тест

print("\nЗадача 5:")
print(f"z-статистика: {z_stat5:.3f}, p-значение: {p_value5:.3f}")
if p_value5 < alpha:
    print("Отвергаем H0: Лекарство повлияло на IQ.")
else:
    print("Не отвергаем H0: Недостаточно доказательств влияния лекарства на IQ.")
```
