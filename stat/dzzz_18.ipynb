```python
from math import comb,exp,factorial

# 1. P(X = 2) для 4 монет
p = 0.5
n = 4
k = 2
P_X2 = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))
print(f"P(X = 2) = {P_X2}")

# 2. P(X >= 1) для 4 монет
P_X0 = (1 - p) ** n
P_X_ge_1 = 1 - P_X0
print(f"P(X >= 1) = {P_X_ge_1}")

# 3. Биномиальное распределение для группы крови А
p_blood = 0.4
n_blood = 4
for k in range(n_blood + 1):
    P = comb(n_blood, k) * (p_blood ** k) * ((1 - p_blood) ** (n_blood - k))
    print(f"P(X = {k}) = {P}")

# 4. Вероятность 8 продаж за час при среднем 10 продаж
lambda_ = 10
k_sales = 8
P_X8 = (lambda_ ** k_sales) * exp(-lambda_) / factorial(k_sales)
print(f"P(X = 8) = {P_X8}")

# 5. Вероятность попасть 10 раз из 12 штрафных бросков
n_shots = 12
k_shots = 10
p_shot = 0.6
P_X10 = comb(n_shots, k_shots) * (p_shot ** k_shots) * ((1 - p_shot) ** (n_shots - k_shots))
print(f"P(X = 10) = {P_X10}")
```
