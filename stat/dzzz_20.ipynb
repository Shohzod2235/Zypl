```python
import numpy as np
import scipy.stats as stats

# ===============================
# Задача 1: Точечные оценки для выборки возраста
# ===============================
# Выборка возрастов
sample_ages = np.array([6, 8, 11, 11, 17, 21, 34, 35, 41, 42, 43, 46, 54, 54, 59, 62])

# Точечная оценка среднего
mean_age = np.mean(sample_ages)

# Точечная оценка стандартного отклонения популяции
# Здесь ddof=0, поскольку мы оцениваем именно популяционное стандартное отклонение
std_age = np.std(sample_ages, ddof=0)

print("Задача 1:")
print("Точечная оценка среднего возраста:", mean_age)
print("Точечная оценка стандартного отклонения:", std_age)


# ===============================
# Задача 2: Вероятность, что доля синих драже будет больше 40%
# ===============================
# В стандартном пакете содержится 17 драже, вероятность того, что драже окажется синим p = 0.24.
# Доля синих > 40% означает, что должно быть как минимум 7 синих драже (7/17 ≈ 41.2%).

n_candies = 17
p_blue = 0.24
# Вероятность того, что будет 6 или меньше синих драже
prob_up_to_6 = stats.binom.cdf(6, n_candies, p_blue)
# Следовательно, вероятность того, что будет 7 или более:
prob_more_than_40 = 1 - prob_up_to_6

print("\nЗадача 2:")
print("Вероятность, что доля синих драже превысит 40%:", prob_more_than_40)


# ===============================
# Задача 3: 90%-й доверительный интервал для среднего веса мужчин
# ===============================
n_men = 64
mean_weight = 84       # Среднее значение выборки (в кг)
std_weight = 24        # Стандартное отклонение выборки (в кг)
alpha = 0.10           # Для 90%-го ДИ

# Для достаточно большого размера выборки используем нормальное распределение
z_value = stats.norm.ppf(1 - alpha/2)
margin_error = z_value * std_weight / np.sqrt(n_men)
ci_lower = mean_weight - margin_error
ci_upper = mean_weight + margin_error

print("\nЗадача 3:")
print("90%-й доверительный интервал для среднего веса мужчин: [{:.2f}, {:.2f}]".format(ci_lower, ci_upper))


# ===============================
# Задача 4: 95%-й доверительный интервал для доли подростков, начавших курить из-за фильмов
# ===============================
n_adolescents = 6522
p_hat = 0.38         # Оценка доли подростков, начавших курить по причине влияния фильмов
alpha = 0.05         # Для 95%-го доверительного интервала

z_95 = stats.norm.ppf(1 - alpha/2)
margin_prop = z_95 * np.sqrt(p_hat * (1 - p_hat) / n_adolescents)
ci_lower_prop = p_hat - margin_prop
ci_upper_prop = p_hat + margin_prop

print("\nЗадача 4:")
print("95%-й доверительный интервал для доли подростков, начавших курить из-за просмотра курящих актёров:")
print("[{:.4f}, {:.4f}]".format(ci_lower_prop, ci_upper_prop))


# ===============================
# Задача 5: 95%-й доверительный интервал для среднего роста женщин
# ===============================
n_women = 49
mean_height = 164     # Средний рост выборки (в см)
std_height = 20       # Стандартное отклонение выборки (в см)
df_t = n_women - 1    # Степени свободы для t-распределения
alpha = 0.05        # Для 95%-го доверительного интервала

t_value = stats.t.ppf(1 - alpha/2, df_t)
margin_height = t_value * std_height / np.sqrt(n_women)
ci_lower_height = mean_height - margin_height
ci_upper_height = mean_height + margin_height

print("\nЗадача 5:")
print("95%-й доверительный интервал для среднего роста женщин: [{:.2f}, {:.2f}]".format(ci_lower_height, ci_upper_height))
```
