```python
from scipy.stats import norm

# Задача 1: Вероятность P(Z <= 1.27)
z1 = 1.27
prob1 = norm.cdf(z1)
print(f"1. P(Z <= 1.27) = {prob1:.4f}")

# Задача 2: Вероятность P(-1.76 < Z < -1.19)
z2_lower = -1.76
z2_upper = -1.19
prob2 = norm.cdf(z2_upper) - norm.cdf(z2_lower)
print(f"2. P(-1.76 < Z < -1.19) = {prob2:.4f}")

# Задача 3: Нахождение среднего значения и стандартного отклонения веса девушек
lower_weight = 38
upper_weight = 52
probability = 0.9545

# Нахождение z-значений для 95.45% доверительного интервала
z_lower = norm.ppf((1 - probability) / 2)
z_upper = norm.ppf(1 - (1 - probability) / 2)

# Нахождение среднего значения и стандартного отклонения
mean_weight = (lower_weight + upper_weight) / 2
std_dev = (upper_weight - lower_weight) / (z_upper - z_lower)
print(f"3. Среднее значение веса: {mean_weight:.2f} кг")
print(f"   Стандартное отклонение: {std_dev:.2f} кг")

# Задача 4: Вероятность длины указательного пальца в диапазоне от 7.3 см до 8.7 см
mu_finger = 7.5
sigma_finger = 0.5
lower_length = 7.3
upper_length = 8.7

prob4 = norm.cdf(upper_length, mu_finger, sigma_finger) - norm.cdf(lower_length, mu_finger, sigma_finger)
print(f"4. P(7.3 < X < 8.7) = {prob4:.4f}")

# Задача 5: Вероятность того, что IQ будет выше 140
mu_iq = 100
sigma_iq = 15
iq = 140

prob5 = 1 - norm.cdf(iq, mu_iq, sigma_iq)
print(f"5. P(X > 140) = {prob5:.4f}")
```
